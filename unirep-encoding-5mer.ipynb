```python
from jax_unirep import get_reps
import numpy as np
```

```python
import jax
jax.devices()
```

```python
def parse_pagg(infile):
    """parse library of PagG library from 10.1021/acscatal.0c00623
    form text file"""
    lib = []
    with open(infile,'r') as handle:
        while True:
            line = handle.readline().split()
            if not line:
                break
            lib.append(line[1][:6])
    return(np.array(lib))
```

```python
# library from PagG paper 10.1021/acscatal.0c00623
pagg_lib = parse_pagg('pagg.txt')
# np.save('pagg_lib', pagg_lib)
```

```python
# generate PagG library encoding and save to disk
h_avg, h_final, c_final = get_reps(pagg_lib)

np.save('unirep_pagg_lib', h_avg)
```
